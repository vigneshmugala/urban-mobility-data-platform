```python
from pyspark.sql import functions as f
source_path = "/Volumes/urban-mobility-data-platform-dev/source/data/yellow_tripdata_2026-01.parquet"
df = spark.read.parquet(source_path)
```

```python
display(df.limit(20))
```

```python
df.printSchema()
```

```python
print("Rows:", df.count())
print("Columns:", len(df.columns))

print(df.columns)
```

```python
display(
  df.select(
   f.min("Airport_fee").alias("min_airport_fee"),
   f.min("DOLocationID").alias("min_DOLocationID"),
   f.min("PULocationID"),
   f.min("RatecodeID"),
   f.min("VendorID"),
   f.min("cbd_congestion_fee"),
   f.min("congestion_surcharge"),
   f.min("extra"),
   f.min("fare_amount"),
   f.min("improvement_surcharge"),
   f.min("mta_tax"),
   f.min("passenger_count"),
   f.min("payment_type"),
   f.min("store_and_fwd_flag"),
   f.min("tip_amount"),
   f.min("tolls_amount"),
   f.min("total_amount"),
   f.min("tpep_dropoff_datetime"),
   f.min("tpep_pickup_datetime"),
   f.min("trip_distance")
  )
)
```

```python
display(
  df.select(
   f.max("Airport_fee").alias("max_airport_fee"),
   f.max("DOLocationID").alias("max_DOLocationID"),
   f.max("PULocationID"),
   f.max("RatecodeID"),
   f.max("VendorID"),
   f.max("cbd_congestion_fee"),
   f.max("congestion_surcharge"),
   f.max("extra"),
   f.max("fare_amount"),
   f.max("improvement_surcharge"),
   f.max("mta_tax"),
   f.max("passenger_count"),
   f.max("payment_type"),
   f.max("store_and_fwd_flag"),
   f.max("tip_amount"),
   f.max("tolls_amount"),
   f.max("total_amount"),
   f.max("tpep_dropoff_datetime"),
   f.max("tpep_pickup_datetime"),
   f.max("trip_distance")
  )
)
```

```python
null_counts = df.select([f.count(f.when(f.col(c).isNull(),c)).alias(c) for c in df.columns])
display(null_counts)
```

```python
display(df.filter((f.col("passenger_count") <= 0) | f.col("passenger_count").isNull()))
```

```python
display(
    df.filter(
        (f.col("trip_distance") <= 0) |
        f.col("trip_distance").isNull()
    ).limit(20)
)
```

```python
display(
    df.filter(
        (f.col("trip_distance") <= 0) |
        f.col("trip_distance").isNull()
    ).limit(20)
)
```

```python
display(
    df.filter(
        f.col("tpep_dropoff_datetime") <
        f.col("tpep_pickup_datetime")
    ).limit(20)
)
```

```python
# Check Duplicates
original_count = df.count()
dedup_count = df.dropDuplicates().count()
print("Original:", original_count)
print("After exact deduplication:", dedup_count)
print("Potential exact duplicates:", original_count - dedup_count)
```
